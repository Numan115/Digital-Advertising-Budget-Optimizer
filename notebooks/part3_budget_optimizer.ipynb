## Part 3: Marketing Budget Recommendation Engine
### Allocating the Additional Rs. 1,000,000 Advertising Budget

### 1. Executive Problem Statement
Company executive leadership has approved an additional **Rs. 1,000,000** for marketing expansion.
Management wants to know:
> *"Which advertising platforms should receive this budget, in what exact proportion, and what is our expected return?"*

### 2. Multi-Criteria Optimization Methodology
Instead of a naive "winner-takes-all" approach, our Recommendation Engine utilizes multi-criteria weighted scoring with portfolio diversification constraints:
- **50% ROAS Weight**: Maximizes revenue generated per rupee spent.
- **30% CPA Efficiency Weight**: Prioritizes platforms with lower customer acquisition costs.
- **20% Conversion Rate Weight**: Favors platforms with proven conversion efficiency.
- **Business Rule Constraint**: Channels with historical ROAS below 3.5x (e.g. YouTube) are paused for direct-response budgets and redirected to organic/branding channels.

```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append('../src')
from budget_optimizer import optimize_budget

results = optimize_budget('../data/processed/advertising_data_with_kpis.csv', additional_budget=1000000.0)
```

### 3. Recommended Budget Distribution Analysis

```python
plt.figure(figsize=(8, 8))
alloc_df = results[results['allocated_budget'] > 0]
colors = ['#2b5c8f', '#4ca585', '#d4a373', '#e76f51']
plt.pie(alloc_df['allocated_budget'], labels=alloc_df['platform'], autopct='%1.1f%%',
        startangle=140, colors=colors, explode=[0.04, 0.04, 0.04, 0.04],
        wedgeprops=dict(width=0.6))
plt.title('Recommended Rs. 1,000,000 Budget Distribution', fontsize=15, fontweight='bold')
plt.show()
```

### 4. Projected Financial Returns & Conversions

```python
fig, ax = plt.subplots(figsize=(10, 6))
results[results['allocated_budget'] > 0].plot(x='platform', y=['allocated_budget', 'proj_revenue', 'proj_net_profit'], 
                                              kind='bar', ax=ax, color=['#3a86ff', '#2a9d8f', '#38b000'])
plt.title('Projected Financial Impact by Platform (PKR)', fontsize=14, fontweight='bold')
plt.ylabel('Amount (PKR)')
plt.xticks(rotation=0)
plt.show()
```
